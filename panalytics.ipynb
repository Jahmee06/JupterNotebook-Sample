```python
import matplotlib.pyplot as plt
import pandas as pd

df['Date'] = df['Date'].astype(str)
df['Date'] = pd.to_datetime(df['Date'])
sales_by_month = df.groupby(df['Date'].dt.to_period("M"))['Total Amount'].sum()
sales_by_month_avg = df.groupby(df['Date'].dt.to_period("M"))['Total Amount'].mean()

plt.figure(figsize=(14, 7))
plt.plot(sales_by_month.index.to_timestamp(), sales_by_month.values, label="Monthly Sales")
plt.plot(sales_by_month_avg.index.to_timestamp(), sales_by_month_avg.values, color="Red", label="Average Monthly Sales")

plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.title('Total Sales by Month')
plt.legend()
plt.show()
```

```python
import matplotlib.dates as mdates
df["Date"]= pd.to_datetime(df["Date"])
df["Week"] = df["Date"].dt.to_period("W")

weekly_sales = df.groupby(df['Week'])['Total Amount'].sum()
weekly_sales_avg = df.groupby(df['Week'])['Total Amount'].mean()

plt.figure(figsize=(10, 4))
plt.plot(weekly_sales.index.to_timestamp(), weekly_sales.values, label="Weekly Sales")
plt.plot(weekly_sales_avg.index.to_timestamp(), weekly_sales_avg.values, color="Red", label="Average Weekly Sales")

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%W'))
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.title('Total Sales by Week')
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
# Reload from the original source
df = pd.read_csv("retail_sales_dataset.csv")

# Clean up column names (remove spaces or weird characters)
df.columns = df.columns.str.strip()

# Check columns
print(df.columns)
```

```python
print(df.columns.tolist())
#Had to find out what I was working with since the weekly cell messed up the Date for me
print(df.isnull().sum()) # SHows missing data
```

```python
# Missing data, this the FIX
df['Gender'] = df['Gender'].str.strip()
df['Gender'] = df['Gender'].str.lower()
df['Gender'].unique()
```

```python
# Date to Week and Gender to Binary
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month

#BINAARYYY
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
```

```python
import pandas as pd

# Define features and targets
X = df[['Age','Month', 'Quantity', 'Price per Unit']]
Y = df['Total Amount']


# Marix x, all predictor variables
# Matrix y, target variable (What we want predicted)
print("Feature matrix shape:", X.shape)
print("Target vector shape:", Y.shape)
```

```python
#Train 80, Test 20
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

#Different methods to seeing set size
print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])

print(f"training set size: {len(X_train)} samples")
print(f"testing set size: {len(X_test)} samples")
```

Building and training the model

```python
model = LinearRegression()

model.fit(X_train,Y_train)
print(model.coef_)
print(model.intercept_)
```

```python
#MAke predictions
y_pred = model.predict(X_test)
comparsion = pd.DataFrame({'Actual': Y_test, 'Predicted': y_pred})
print(comparsion)
```
